```python
age = 17

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
def age_category(age):
    if age >= 18:
        return "Adult"
    else:
        return "Minor"

df["Age_Category"] = df["age"].apply(age_category)
```

```python
df[["age", "Age_Category"]].head()
```

```python
def age_group(age):
    if age <= 16:
        return "Young"
    elif age <= 18:
        return "Teen"
    else:
        return "Older"

df["Age_Group"] = df["age"].apply(age_group)
```

```python
df[["age", "Age_Group"]].head()
```

```python
def student_category(age, medu):
    if age >= 18 and medu >= 3:
        return "Category A"
    else:
        return "Category B"

df["Student_Category"] = df.apply(
    lambda row: student_category(row["age"], row["Medu"]),
    axis=1
)
```

```python
def location_category(address):
    if address == "U" or address == "R":
        return "Known Location"
    else:
        return "Other"

df["Location_Category"] = df["address"].apply(location_category)
```

```python
def mother_job_category(job):
    if job == "teacher":
        return "Education"
    elif job == "health":
        return "Healthcare"
    else:
        return "Other"

df["Mother_Job_Category"] = df["Mjob"].apply(mother_job_category)
```

```python
def education_comparison(medu, fedu):
    if medu > fedu:
        return "Mother education higher"
    elif medu < fedu:
        return "Father education higher"
    else:
        return "Same education level"

df["Parent_Education"] = df.apply(
    lambda row: education_comparison(row["Medu"], row["Fedu"]),
    axis=1
)
```
